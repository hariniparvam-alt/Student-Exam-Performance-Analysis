```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_exam_performance.csv")

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
df.describe()
```

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

print(missing_values.sort_values(ascending=False))
```

```python
# Check data types of columns with missing values
columns_to_check = [
    "attendance_percentage",
    "time_management_score",
    "notes_quality",
    "previous_gpa",
    "sleep_quality",
    "parent_education",
    "device_availability"
]

print(df[columns_to_check].dtypes)
```

```python

print("Parent education:")
print(df["parent_education"].value_counts(dropna=False).head(10))

print("\nDevice availability:")
print(df["device_availability"].value_counts(dropna=False).head(10))

print("\nNotes quality:")
print(df["notes_quality"].value_counts(dropna=False).head(10))

print("\nSleep quality:")
print(df["sleep_quality"].value_counts(dropna=False).head(10))

```

```python

# Create a copy for data cleaning
df_clean = df.copy()

# Numerical columns: fill missing values with median
numerical_columns = [
    "attendance_percentage",
    "time_management_score",
    "previous_gpa"
]

for col in numerical_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Categorical columns: fill missing values with Unknown
categorical_columns = [
    "notes_quality",
    "sleep_quality",
    "parent_education",
    "device_availability"
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

print("Missing values handled successfully!")
```

```python

print("Missing values before cleaning:", df.isnull().sum().sum())
print("Missing values after cleaning:", df_clean.isnull().sum().sum())
 
```

```python
# Check completely duplicated rows
print("Duplicate rows:", df_clean.duplicated().sum())

# Check repeated student IDs
print("Duplicate student IDs:", df_clean["student_id"].duplicated().sum())
```

```python

# Check numerical ranges
print("Age range:")
print(df_clean["age"].min(), "to", df_clean["age"].max())

print("\nExam score range:")
print(df_clean["exam_score"].min(), "to", df_clean["exam_score"].max())

print("\nAttendance percentage range:")
print(
    df_clean["attendance_percentage"].min(),
    "to",
    df_clean["attendance_percentage"].max()
)

print("\nPrevious GPA range:")
print(
    df_clean["previous_gpa"].min(),
    "to",
    df_clean["previous_gpa"].max()
)

print("\nPass status values:")
print(df_clean["pass_status"].value_counts())
```

```python
df_clean.to_csv("student_exam_performance_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Dataset shape:", df_clean.shape)
```

```python

# Select numerical columns
numeric_df = df_clean.select_dtypes(include="number")

# Calculate correlations with exam score
correlations = numeric_df.corr()["exam_score"].drop("exam_score")

# Display the strongest relationships first
correlations = correlations.sort_values(
    ascending=False,
    key=abs
)

print("Factors most strongly related to exam scores:")
print(correlations.head(15))
```

```python

import matplotlib.pyplot as plt
import pandas as pd

# Group students by daily study hours
bins = [0, 2, 4, 6, 8, float("inf")]
labels = ["0–2 hours", "2–4 hours", "4–6 hours",
          "6–8 hours", "8+ hours"]

df_clean["study_hours_group"] = pd.cut(
    df_clean["study_hours_per_day"], 
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Calculate average exam score for each group
avg_scores = df_clean.groupby(
    "study_hours_group", observed=False
)["exam_score"].mean()

print(avg_scores)

# Create a bar chart
avg_scores.plot(kind="bar", figsize=(8, 5))

plt.title("Average Exam Score by Daily Study Hours")
plt.xlabel("Daily Study Hours")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Calculate correlation
correlation = df_clean[
    ["previous_exam_score", "exam_score"]
].corr().iloc[0, 1]

print("Correlation:", round(correlation, 3))

# Create a scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df_clean["previous_exam_score"],
    df_clean["exam_score"],
    alpha=0.3
)

plt.title("Previous Exam Score vs Current Exam Score")
plt.xlabel("Previous Exam Score")
plt.ylabel("Current Exam Score")
plt.tight_layout()
plt.show()
```

```python

# Calculate correlation
attendance_corr = df_clean[
    ["attendance_percentage", "exam_score"]
].corr().iloc[0, 1]

print("Attendance vs Exam Score Correlation:",
      round(attendance_corr, 3))

# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df_clean["attendance_percentage"],
    df_clean["exam_score"],
    alpha=0.3
)

plt.title("Attendance Percentage vs Exam Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python

# Calculate correlation
stress_corr = df_clean[
    ["stress_level", "exam_score"]
].corr().iloc[0, 1]

print("Stress vs Exam Score Correlation:",
      round(stress_corr, 3))

# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df_clean["stress_level"],
    df_clean["exam_score"],
    alpha=0.3
)

plt.title("Stress Level vs Exam Score")
plt.xlabel("Stress Level")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python

# Calculate average exam score by sleep quality
sleep_scores = df_clean.groupby(
    "sleep_quality"
)["exam_score"].mean().sort_values(ascending=False)

print("Average exam score by sleep quality:")
print(sleep_scores.round(2))

# Create bar chart
sleep_scores.plot(kind="bar", figsize=(8, 5))

plt.title("Average Exam Score by Sleep Quality")
plt.xlabel("Sleep Quality")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Key Performance Indicators (KPIs)

total_students = len(df_clean)
average_score = df_clean["exam_score"].mean()
pass_count = (df_clean["pass_status"] == "Pass").sum()
fail_count = (df_clean["pass_status"] == "Fail").sum()
pass_rate = (pass_count / total_students) * 100

print("Total Students:", total_students)
print("Average Exam Score:", round(average_score, 2))
print("Students Passed:", pass_count)
print("Students Failed:", fail_count)
print("Pass Rate:", round(pass_rate, 2), "%")
```

```python

# Save correlation results
correlation_df = correlations.reset_index()
correlation_df.columns = ["Factor", "Correlation"]

correlation_df.to_csv("exam_score_correlations.csv", index=False)

print("Correlation report saved successfully!")
```
